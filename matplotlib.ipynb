```python
x= np.linspace(0,5,11)
x
y= x**2

y
```

```python
plt.plot(x,y)
plt.title("Basic")
plt.xlabel("X- Axis")
plt.ylabel("Y-Axis")
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x**3,x)
plt.subplot(2,2,4)
plt.plot(x**2,y**5)
```

**Object Oriented Way**

```python
fig = plt.figure(dpi = 2000)
axis1 = fig.add_axes([0,0,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.15,0.3,.3,.3])
axis2.plot(x,y)
```
